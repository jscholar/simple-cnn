```python
import torch
from torch import nn
```

```python
## A convolution neural network is one that uses learned filters that operate on subsets of the input data.
## For (an oversimplified) example, a specific filter can be trained to detect circles, and is applied to 4x4 pixel subsets of the input image.

conv_ex = nn.Conv2d(
    in_channels=3, ## Like RGB
    out_channels=16, ## The number of filters/channels.
    kernel_size=3, ## This is the size of the subsection of the filter.
    padding=1
)

print(conv_ex.weight.shape)
print(conv_ex.bias.shape)
```

```python
## Tweak this and see what happens?
## These represent the number of filters/channels learned in each convolution layer

CONV_LAYER_1_CHANNELS = 16
CONV_LAYER_2_CHANNELS = 32
```

```python
class TennisBallCNN(nn.Module):
  conv_1_channels = CONV_LAYER_1_CHANNELS
  conv_2_channels = CONV_LAYER_2_CHANNELS

  def __init__(self):
    super().__init__()

    # [3, 64, 64] -> [conv_1_channels, 32, 32]
    first_conv = nn.Sequential(
      nn.Conv2d(3, self.conv_1_channels, kernel_size=3, padding=1),
      nn.ReLU(),
      nn.MaxPool2d(2),
    )

    # [3, 64, 64] -> [conv_2_channels, 16, 16]
    second_conv = nn.Sequential(
      nn.Conv2d(self.conv_1_channels, self.conv_2_channels, kernel_size=3, padding=1),
      nn.ReLU(),
      nn.MaxPool2d(2),
    )

    self.network = nn.Sequential(
      first_conv,

      second_conv,

      nn.Flatten(),

      nn.Linear(self.conv_2_channels * 16 * 16, 64),
      nn.ReLU(),

      # [64] -> [2]
      nn.Linear(64, 2),
    )

  def forward(self, x):
    return self.network(x)
```
